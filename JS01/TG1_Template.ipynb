## Bagian 1: Instalasi dan Penjelasan Library

### 1. Instalasi PyPREP

```python
!pip install pyprep
```

**Kegunaan PyPREP:**

PyPREP adalah Python implementation dari PREP (Preprocessing Pipeline) yang digunakan untuk preprocessing data EEG (Electroencephalography). Library ini membantu dalam:
- Mendeteksi dan memperbaiki bad channels dalam data EEG
- Re-referencing data EEG
- Standardisasi preprocessing data neuroimaging
- Implementasi pipeline PREP yang robust untuk analisis sinyal otak

### 2. Instalasi Scipy

```python
!pip install scipy
```

**Kegunaan Scipy:**

Scipy (Scientific Python) adalah library fundamental untuk komputasi ilmiah di Python. Kegunaan utamanya:
- Optimasi algoritma (optimization)
- Integrasi dan diferensiasi numerik
- Interpolasi data
- Pemrosesan sinyal (signal processing)
- Aljabar linier (linear algebra)
- Statistik dan distribusi probabilitas
- FFT (Fast Fourier Transform) untuk analisis frekuensi

### 3. Instalasi wandb

```python
!pip install wandb
```

**Kegunaan wandb (Weights & Biases):**

Wandb adalah platform untuk tracking, visualisasi, dan kolaborasi dalam eksperimen machine learning:
- Tracking hyperparameter dan metrics model
- Visualisasi training progress secara real-time
- Membandingkan hasil berbagai eksperimen
- Menyimpan dan versioning model
- Kolaborasi tim dalam proyek ML
- Monitoring resource usage (GPU, CPU, memory)
- Reproducibility eksperimen ML

### 4. Instalasi pyECG

```python
!pip install py-ecg-detectors
```

**Kegunaan pyECG:**

pyECG adalah library untuk deteksi dan analisis sinyal ECG (Electrocardiography):
- Deteksi QRS complex dalam sinyal ECG
- Ekstraksi heart rate dan heart rate variability
- Implementasi berbagai algoritma deteksi gelombang ECG
- Preprocessing sinyal jantung
- Analisis ritme jantung untuk aplikasi kesehatan
- Berguna untuk biomedical signal processing
