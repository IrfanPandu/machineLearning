# JS02 - Praktikum 3: Seleksi Fitur, Encoding, dan Standardisasi

Dataset: Titanic (Fixed)

## Import Library

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

## Langkah 1 - Load Data

```python
dpath = 'Titanic-Dataset-fixed.csv'
df = pd.read_csv(dpath)
df.head()
```

## Langkah 2 - Slice Data (Seleksi Fitur)

```python
df = df[['Survived', 'Pclass', 'Age', 'Sex', 'Cabin']]
df.head()
```

## Langkah 3 - Encoding

```python
le = LabelEncoder()
df['Sex'] = le.fit_transform(df['Sex'])
df['Cabin'] = le.fit_transform(df['Cabin'])
```

## Langkah 4 - Verifikasi Hasil Encoding

```python
df.head()
```

## Langkah 5 - Standardisasi

```python
std = StandardScaler()
df['Age'] = std.fit_transform(df[['Age']])
```
