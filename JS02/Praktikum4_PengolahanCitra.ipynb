```python
import cv2
import matplotlib.pyplot as plt
```

## Langkah 1 - Load Gambar

```python
img = cv2.imread("Lenna.png")

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.title("Gambar Asli")
plt.axis("off")
plt.show()
```

## Langkah 2 - Resize Gambar

```python
img_resized = cv2.resize(img_rgb, (128, 128))

plt.imshow(img_resized)
plt.title("Gambar setelah Resize (128x128)")
plt.axis("off")
plt.show()
```

## Langkah 3 - Ubah ke Grayscale

```python
img_gray = cv2.cvtColor(img_resized, cv2.COLOR_RGB2GRAY)

plt.imshow(img_gray, cmap="gray")
plt.title("Gambar Grayscale")
plt.axis("off")
plt.show()
```

## Bonus: Perbandingan Side by Side

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].imshow(img_rgb)
axes[0].set_title("Asli")
axes[0].axis("off")

axes[1].imshow(img_resized)
axes[1].set_title("Resize (128x128)")
axes[1].axis("off")

axes[2].imshow(img_gray, cmap="gray")
axes[2].set_title("Grayscale")
axes[2].axis("off")

plt.tight_layout()
plt.show()
```
