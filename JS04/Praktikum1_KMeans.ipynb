# JS04 - Praktikum 1: KMeans Clustering

Dataset: Iris

## Persiapan Data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

df = pd.read_csv('Iris.csv')

df.head()
```

## Seleksi Fitur

```python
X = df.iloc[:, 1:-1]
y = df.iloc[:, -1]

X.head()
```

## Plot Data

```python
plt.scatter(X.iloc[:, 0], X.iloc[:, 1], s = 100)
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.title('Data Iris - Sepal Length vs Sepal Width')
plt.show()
```

## Buat Model KMeans (k=2)

```python
cl_kmeans = KMeans(n_clusters=2)

y_kmeans = cl_kmeans.fit_predict(X)
```

## Plot Hasil Cluster

```python
plt.scatter(X.iloc[:, 0], X.iloc[:, 1], s = 100, c=y_kmeans)

centers = cl_kmeans.cluster_centers_
plt.scatter(centers[:, 0], centers[:, 1], c='red', s=200, alpha=0.5, marker='X')
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.title('Hasil Clustering KMeans (k=2)')
plt.show()
```

## Cek Nilai SSE

```python
print(f'Nilai SSE: {cl_kmeans.inertia_}')
```

## Implementasi Metode Elbow

```python
sse = []

K = range(1,10)

for k in K:
    kmeanModel = KMeans(n_clusters=k)
    kmeanModel.fit(X)
    sse.append(kmeanModel.inertia_)

plt.figure(figsize=(8,4))
plt.plot(K, sse, "bx-")
plt.xlabel("k")
plt.ylabel("SSE")
plt.title("Metode Elbow untuk Mengetahui Jumlah k Terbaik")
plt.grid(True)
plt.show()
```

## Cek Nilai SSE Setiap k

```python
for idx, sse_val in enumerate(sse, start=1):
    print(f'k={idx}; SSE={sse_val}')
```

## Model dengan k Optimal (k=3)

```python
cl_kmeans_optimal = KMeans(n_clusters=3)
y_kmeans_optimal = cl_kmeans_optimal.fit_predict(X)

plt.scatter(X.iloc[:, 0], X.iloc[:, 1], s = 100, c=y_kmeans_optimal, cmap='viridis')
centers = cl_kmeans_optimal.cluster_centers_
plt.scatter(centers[:, 0], centers[:, 1], c='red', s=200, alpha=0.5, marker='X')
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.title('Hasil Clustering KMeans (k=3 - Optimal)')
plt.show()

print(f'Nilai SSE dengan k=3: {cl_kmeans_optimal.inertia_}')
```
