Penjelasan Library

PyPREP

Merupakan library python yang digunakan untuk melakukan preprocessing data EEG / Electroencephalography. Library ini membantu meningkatkan kualitas data dengan mengidentifikasi dan menangani gangguan atau noise sebelum data digunakan dalam proses analisis maupun pembelajaran mesin.

SciPy

SciPy merupakan library Python untuk komputasi ilmiah yang menyediakan berbagai fungsi matematis dan numerik. Library ini mendukung kebutuhan seperti statistik, optimasi, aljabar linear, integrasi, serta pengolahan sinyal yang banyak digunakan dalam analisis data dan pembelajaran mesin.

Weights & Biases (W&B)

merupakan platform yang digunakan untuk melakukan experiment tracking dalam pengembangan model pembelajaran mesin. W&B memungkinkan peneliti atau pengembang mencatat hyperparameter, metrik performa, konfigurasi eksperimen, serta hasil pelatihan model secara terstruktur sehingga proses evaluasi dan perbandingan eksperimen menjadi lebih mudah.

PyECG

merupakan library Python yang digunakan untuk pengolahan dan pengelolaan data ECG (Electrocardiography). Library ini mendukung pembacaan dan penulisan data rekaman ECG dari beberapa format dataset sehingga dapat digunakan sebagai bagian dari proses analisis sinyal dan penelitian berbasis data elektrokardiografi.

1. Pelanggaran Etika dan Hukum dalam Penggunaan AI

Penggunaan AI dapat menimbulkan pelanggaran etika dan hukum apabila digunakan tanpa memperhatikan hak, privasi, dan kepentingan pihak lain. Salah satu contohnya adalah penggunaan teknologi deepfake untuk memanipulasi wajah atau suara seseorang tanpa izin, kemudian menyebarkannya seolah-olah merupakan informasi yang benar. Tindakan tersebut dapat merugikan pihak yang menjadi objek manipulasi, terutama jika digunakan untuk penipuan, pencemaran nama baik, atau penyebaran informasi palsu. Oleh karena itu, penggunaan AI perlu memperhatikan aspek etika, privasi, hak cipta, serta peraturan yang berlaku agar teknologi tersebut digunakan secara bertanggung jawab.

2. Dampak Energi dan Lingkungan dari AI

Pengembangan dan penggunaan AI membutuhkan sumber daya komputasi yang cukup besar, terutama pada proses pelatihan dan pengoperasian model dengan skala besar. Kebutuhan komputasi tersebut menyebabkan peningkatan konsumsi listrik pada pusat data serta membutuhkan sistem pendingin yang juga menggunakan energi dan air. Selain itu, produksi perangkat keras seperti server dan GPU membutuhkan bahan baku serta dapat menghasilkan limbah elektronik. Oleh karena itu, pengembangan AI perlu mempertimbangkan efisiensi model, penggunaan perangkat keras secara optimal, serta pemanfaatan sumber energi yang lebih ramah lingkungan untuk mengurangi dampak terhadap lingkungan.

3. Upaya Mengurangi Dampak Lingkungan AI

Dampak lingkungan dari penggunaan AI dapat dikurangi melalui beberapa pendekatan, seperti menggunakan model yang lebih efisien, mengoptimalkan proses pelatihan, serta menghindari penggunaan sumber daya komputasi yang tidak diperlukan. Penggunaan energi terbarukan pada pusat data dan pemanfaatan perangkat keras secara berkelanjutan juga dapat membantu mengurangi emisi dan limbah elektronik. Dengan menerapkan efisiensi tersebut, manfaat AI tetap dapat dikembangkan tanpa mengabaikan aspek keberlanjutan lingkungan.
